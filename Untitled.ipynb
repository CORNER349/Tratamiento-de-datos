```python
import pandas as pd
import numpy as np
dataset=pd.read_csv("datos333respaldo.csv")
print(dataset.head())
```

```python
dataset.shape
```

```python
print(dataset.isnull().sum())
```

```python
# dataset with missed values
dataset["Entero"][:10]
```

```python
dataset.dropna(inplace=True)
```

```python
print(dataset.isnull().sum())
```

```python
dataset.shape
```

```python
dataset["Entero"]=dataset["Entero"].replace(np.nan, dataset["Entero"].mean())
```

```python
print(dataset["Entero"][:10])
```

```python
dataset["Decimal"]=dataset["Decimal"].replace(np.nan,dataset["Decimal"].median())
```

```python
print(dataset["Decimal"][:10])
```

```python
print(dataset.isnull().sum())
```

```python
import statistics
dataset["Texto"]=dataset["Texto"].replace(np.nan,statistics.mode(dataset["Texto"]))
```

```python
print(dataset.isnull().sum())
```

```python
import statistics
dataset["Peso"]=dataset["Peso"].replace(np.nan,statistics.mode(dataset["Peso"]))
```

```python
print(dataset.isnull().sum())
```

```python
dataset.isnull().sum()
```

```python
dataset["Texto"]=dataset["Texto"].fillna('Texto')
```
