```python
import numpy as np 
import pandas as pd 
```

```python
data = pd.read_csv("datos33_rellenado.csv")
print(data.head(20))
print(data["IMC"].median())
print(data["IMC"].head(20))
```

```python
columna=data["IMC"]
limiteinferior=18.5
limitesuperior=25
atipicos=columna[(columna<limiteinferior) | (columna>limitesuperior)]
print(atipicos)
print(len(atipicos))
```

```python
print(data["IMC"].head(20))
```

```python
mediana=data["IMC"].median()
print(mediana)
medianaaltura=data["Altura"].median()
print(medianaaltura)
medianapeso=data["Peso"].median()
print(medianapeso)
data.loc[atipicos.index,"IMC"]=mediana
data.loc[atipicos.index,"Altura"]=medianaaltura
data.loc[atipicos.index,"Peso"]=medianapeso
```
